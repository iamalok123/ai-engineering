### Short-Term vs Long-Term

```python
# Short-term
# Current conversation

# Example:
# User: My name is Alok.
# User: What is my name?




# Long-term
# Information worth remembering across conversations

# Example:

# User prefers Python.
# User prefers Gemini.
# User is building StudyFlow.
# User's preferred response style is concise.




# Current LangChain documentation explicitly separates short-term thread memory from 
# long-term memory that persists across threads/sessions.
```

### Embeddings

```python
# Suppose:

# "My favorite programming language is Python."

# An embedding model converts it into a vector:

# [0.021, -0.182, 0.743, ...]



# Another query:

# "What language does the user prefer?"

# is also converted into a vector.



# Then:

# query vector
#       ↓
# similarity search
#       ↓
# "My favorite programming language is Python."

# That's semantic memory.
```

### Gemini Embeddings with LangChain

```python
from langchain_google_genai import GoogleGenerativeAIEmbeddings

embeddings = GoogleGenerativeAIEmbeddings(
    model="gemini-embedding-2"
)

vector = embeddings.embed_query(
    "My favorite programming language is Python."
)

print(len(vector))
print(vector[:5])
```

```python
# You now have:

# text
#  ↓
# Gemini Embedding 2
#  ↓
# vector
```

### Simple Vector Memory

```python
from langchain_core.vectorstores import InMemoryVectorStore

vector_store = InMemoryVectorStore(embeddings)


# Store memories:

memories = [
    "The user prefers Python.",
    "The user is learning LangChain.",
    "The user is building AI applications.",
    "The user prefers concise technical explanations."
]

vector_store.add_texts(memories)



# Search:

results = vector_store.similarity_search(
    "What programming language does the user like?",
    k=2
)

for doc in results:
    print(doc.page_content)



# The important architecture is:

#            Long-term memory

# User information
#       ↓
# Embedding
#       ↓
# Vector store
#       ↓
# Similarity search
#       ↓
# Relevant memories
#       ↓
# Prompt
#       ↓
# Gemini
```

### Don't Store Everything in Vector Memory

```python
This is an important production concept.




Bad:

Every single chat message
        ↓
vector DB




Better:

Conversation
     ↓
decide what is worth remembering
     ↓
important facts/preferences
     ↓
long-term store





For example:

User:
"Okay, thanks!"

❌ Don't store.

User:
"I prefer TypeScript for frontend projects."

✅ Potential memory.
```
