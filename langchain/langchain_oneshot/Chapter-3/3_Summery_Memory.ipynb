### Summary Memory?

```python
# Suppose:

# 100 messages



# Full history:

# 100 messages → Gemini



# Instead:

# 100 messages
#       ↓
# summary
#       ↓
# "User is Alok, a CS student.
# He is learning GenAI and prefers
# Gemini and LangChain..."




# Then:

# summary + recent messages
#           ↓
# Gemini

# Much smaller context.
```

### Modern Summarization Middleware

```python
from langchain.agents import create_agent
from langchain.agents.middleware import SummarizationMiddleware
from langgraph.checkpoint.memory import InMemorySaver

agent = create_agent(
    model="google_genai:gemini-3.6-flash",
    tools=[],
    middleware=[
        SummarizationMiddleware(
            model="google_genai:gemini-3.6-flash",
            trigger=("tokens", 4000),
            keep=("messages", 20)
        )
    ],
    checkpointer=InMemorySaver()
)

# Meaning:

# When conversation becomes large
#              ↓
# summarize old messages
#              ↓
# keep recent messages
#              ↓
# continue conversation

# Current LangChain documents this exact strategy using SummarizationMiddleware.
```

### Hybrid Memory

```python
# This is more useful than pure summary memory.

#                     MEMORY
#                        │
#               ┌────────┴────────┐
#               │                 │
#          Old messages       Recent messages
#               │                 │
#           Summary             Full text
#               │                 │
#               └────────┬────────┘
#                        ↓
#                     Gemini

# Example:

# Summary:
# User is building a PDF AI assistant.

# Recent:
# User: I changed the chunk size to 500.
# AI: That's reasonable...

# Current:
# Should I use 700 instead?

# Gemini gets both:

# summary + recent messages + current question

# This is generally a much more useful mental model than memorizing ConversationSummaryMemory.
```
