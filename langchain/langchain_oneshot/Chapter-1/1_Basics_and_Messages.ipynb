## LLM Call

### 1. Normal API call with gemini api

```python
from google import genai

client = genai.Client()

interaction = client.interactions.create(
    model="gemini-3.5-flash-lite",
    input="Explain how AI works in a few words"
)
print(interaction.output_text)
```

### 2. Normal one-shot chat with langchain

```python
from dotenv import load_dotenv
from langchain_google_genai import ChatGoogleGenerativeAI

load_dotenv()

llm = ChatGoogleGenerativeAI(
    model="gemini-3.1-flash-lite",
    temperature=0.6
)

response = llm.invoke(
    "What is LangChain ins short?"
)

# print(response.content)
print(response.text)
```

### 3. Normal chat + streaming

```python
from dotenv import load_dotenv
from langchain_google_genai import ChatGoogleGenerativeAI

load_dotenv()

llm = ChatGoogleGenerativeAI(
    model="gemini-3.5-flash-lite"
)

for chunk in llm.stream( "Explain LangChain in short and simple words." ):
    print(chunk.text, end="", flush=True)
```

### 8. Multiple turns with the same messages list

```python
from langchain_core.messages import (
    SystemMessage,
    HumanMessage,
    AIMessage
)
from langchain_google_genai import ChatGoogleGenerativeAI

llm = ChatGoogleGenerativeAI(
    model="gemini-3.5-flash-lite"
)

messages = [
    SystemMessage(
        content="You are a DSA teacher."
    )
]

messages.append(
    HumanMessage(
        content="What is binary search?"
    )
)

response = llm.invoke(messages)

print("AI:", response.text)

messages.append(
    AIMessage(
        content=response.content
    )
)

messages.append(
    HumanMessage(
        content="What is its time complexity?"
    )
)

response = llm.invoke(messages)

print("AI:", response.text)
```

### 9. A proper simple chatbot loop

```python
from langchain_core.messages import (
    HumanMessage,
    AIMessage
)
from langchain_google_genai import ChatGoogleGenerativeAI

llm = ChatGoogleGenerativeAI(
    model="gemini-3.5-flash-lite"
)

messages = []

while True:

    user_input = input("You: ")

    if user_input.lower() == "exit":
        break

    messages.append(
        HumanMessage(content=user_input)
    )

    response = llm.invoke(messages)

    print("AI:", response.text)
    print("Enter 'exit'  to end the communication.")

    messages.append(
        AIMessage(content=response.content)
    )
```

### 10. Chatbot + streaming

```python
from langchain_core.messages import AIMessage, HumanMessage
from langchain_google_genai import ChatGoogleGenerativeAI

llm = ChatGoogleGenerativeAI(model="gemini-3.5-flash-lite")

messages = []

while True:
    user_input = input("You: ")

    if user_input.lower() == "exit":
        break

    messages.append(HumanMessage(content=user_input))

    print("AI: ", end="")

    full_chunk = None
    for chunk in llm.stream(messages):
        print(chunk.text(), end="", flush=True)
        full_chunk = chunk if full_chunk is None else full_chunk + chunk

    print()

    if full_chunk:
        messages.append(full_chunk)
```

### 11. System prompt + chatbot + streaming

```python
from langchain_core.messages import AIMessage, HumanMessage
from langchain_google_genai import ChatGoogleGenerativeAI

llm = ChatGoogleGenerativeAI(model="gemini-3.5-flash-lite")


messages = [
    SystemMessage(
        content="""
        You are a helpful AI tutor.
        Explain concepts simply.
        Use examples when useful.
        """
    )
]

while True:

    user_input = input("You: ")
    if user_input.lower() == "exit":
        break

    messages.append(
        HumanMessage(content=user_input)
    )

    print("AI: ", end="")

    full_response = ""
    for chunk in llm.stream(messages):
        print(chunk.text, end="", flush=True)
        full_chunk = chunk if full_chunk is None else full_chunk + chunk

    print()

    messages.append(
        AIMessage(content=full_response)
    )
```

### 14. Async chatbot

```python
from langchain_google_genai import ChatGoogleGenerativeAI

llm = ChatGoogleGenerativeAI(model="gemini-3.5-flash-lite")

async def main():
    messages = [
        ("system", "You are a helpful AI tutor."),
        ("human", "Explain FastAPI.")
    ]

    response = await llm.ainvoke(messages)
    print(response.text)

await main()
```

### 15. Async streaming

```python
from langchain_google_genai import ChatGoogleGenerativeAI

llm = ChatGoogleGenerativeAI(model="gemini-3.5-flash-lite")

async def main():
    messages = [
        ("system", "You are a helpful AI tutor."),
        ("human", "Explain FastAPI.")
    ]

    async for chunk in llm.astream(messages):
        print(
            chunk.text,
            end="",
            flush=True
        )


await main()
```
